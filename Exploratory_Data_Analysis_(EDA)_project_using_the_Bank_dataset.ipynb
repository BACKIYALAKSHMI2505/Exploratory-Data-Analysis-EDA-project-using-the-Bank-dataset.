# 5. Outlier Detection

Potential outliers were identified using:

These methods help identify unusual values in numerical columns.

1. IQR method - Interquartile Range

```python
Q1 = data["income"].quantile(0.25)
Q3 = data["income"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = data[
    (data["income"] < lower) |
    (data["income"] > upper)
]

print("Income Outliers:")
display(outliers)
```

 2. Z-score

```python
from scipy.stats import zscore

data["income_zscore"] = zscore(data["income"])

outliers_z = data[
    abs(data["income_zscore"]) > 3
]

display(outliers_z)
```

#6. Data transformation

Label Encoding was used to convert categorical variables into numerical values.

1.  Encoding

```python
from sklearn.preprocessing import LabelEncoder

data_encoded = data.copy()

encoder = LabelEncoder()

categorical_columns = data_encoded.select_dtypes(
    include="object"
).columns

for col in categorical_columns:
    data_encoded[col] = encoder.fit_transform(
        data_encoded[col].astype(str)
    )

display(data_encoded.head())
```

# 7. Feature Scaling

The following feature scaling methods were applied:

These methods help bring numerical features into comparable scales.

 1. Normalization

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

data_normalized = data_encoded.copy()

columns = ["age", "income", "children", "mortgage"]

data_normalized[columns] = scaler.fit_transform(
    data_normalized[columns]
)

display(data_normalized.head())
```

2. Standardization

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

data_standardized = data_encoded.copy()

data_standardized[columns] = scaler.fit_transform(
    data_standardized[columns]
)

display(data_standardized.head())
```
